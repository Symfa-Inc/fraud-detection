```python
from autogluon.tabular import TabularDataset, TabularPredictor

train_data = TabularDataset("../data/train.parquet")
train_data
```

```python
# Only features used by the frontend (top 10 by SHAP importance)
FEATURE_COLS = [
    "annual_income",
    "age_of_driver",
    "claim_day_of_week",
    "high_education_ind",
    "past_num_of_claims",
    "safty_rating",
    "witness_present_ind",
    "gender",
    "claim_est_payout",
    "living_status",
]
train_data = train_data[FEATURE_COLS + ["fraud"]]
train_data
```

```python
label = "fraud"
print(f"Unique classes: {list(train_data[label].unique())}")
```

```python
predictor = TabularPredictor(label=label, path="../models/AutogluonModels/reduced").fit(
    train_data
)
```

```python
test_data = TabularDataset("../data/test.parquet")
test_data
```

```python
test_data = test_data[FEATURE_COLS + ["fraud"]]
test_data
```

```python
y_pred = predictor.predict(test_data)
y_pred
```

```python
y_pred_proba = predictor.predict_proba(test_data)
y_pred_proba
```

```python
predictor.evaluate(test_data)
```
